```python
import pandas as pd
from datetime import datetime
from joblib import Parallel, delayed
import dataset, label  # your dataset.py with train_data()
import calendar_util  # for get_training_dates

def _train_data_task(symbol, date, feature_bar, port_freq, label_bar, local_timezone):
    """
    Helper task to call train_data safely.
    """
    try:
        label_df = label.label_data(symbol, date, port_freq, label_bar, local_timezone)
        train_df = dataset.train_data(symbol, date, port_freq, feature_bar, local_timezone)
        df = train_df.merge(label_df, on = ['timestamp', 'symbol'])
        if df is not None and not df.empty:
            return df
    except Exception as e:
        print(f"[ERROR] train_data failed for {symbol} on {date}: {e}")
    return None

def create_multi_day_dataset(symbols, dates, port_freq, feature_bar, label_bar, local_timezone="Asia/Kolkata", nproc=4):
    """
    Create combined dataset for multiple symbols and dates in parallel using joblib.

    Parameters
    ----------
    symbols : str or list[str]
        Single symbol or list of trading symbols.
    dates : pd.DatetimeIndex, list, or list-like of dates
        Dates to collect data for.
    port_freq : str
        Interval string, e.g. '15min', '30min'.
    local_timezone : str
        Timezone string.
    nproc : int
        Number of parallel workers.

    Returns
    -------
    pd.DataFrame
        Combined DataFrame of all training data.
    """
    # Normalize symbols and dates to lists
    if hasattr(dates, "__iter__") and not isinstance(dates, (str, bytes)):
        dates = [d.date() for d in pd.to_datetime(dates)]
    else:
        dates = [pd.to_datetime(dates).date()]
    if isinstance(symbols, str):
        symbols = [symbols]
    if isinstance(dates, (pd.DatetimeIndex, list, tuple)):
        dates = list(pd.to_datetime(dates).date)
    else:
        # single date input
        dates = [pd.to_datetime(dates).date()]


    # Create all (symbol, date) pairs
    tasks = [(symbol, date, port_freq, feature_bar, label_bar, local_timezone) for symbol in symbols for date in dates]

    results = Parallel(n_jobs=nproc, backend='loky')(
        delayed(_train_data_task)(symbol, date, port_freq, feature_bar, label_bar, local_timezone) for symbol, date, port_freq, feature_bar, label_bar, local_timezone in tasks
    )

    # Filter out None or empty results
    results = [df for df in results if df is not None and not df.empty]

    if not results:
        print("[WARN] No data collected for any symbol-date pair.")
        return pd.DataFrame()

    combined_df = pd.concat(results, ignore_index=True)
    print(f"[INFO] Combined dataset contains {len(combined_df)} rows from {len(symbols)} symbols and {len(dates)} dates.")
    return combined_df


def get_train_data(df, target_date, lookback_days, lag=2, model_train_freq="weekly"):
    """
    Slice training data from df based on lookback window and lag.

    Parameters
    ----------
    df : pd.DataFrame
        The full dataset containing a 'timestamp' column.
    target_date : datetime or str
        The date to predict for (exclusive from training set).
    lookback_days : int
        How many days to look back for training data.
    lag : int
        Gap (in days) between the last training date and target date.
    model_train_freq : str
        Frequency passed to `calendar_util.get_training_dates`.

    Returns
    -------
    pd.DataFrame
        The filtered training dataset.
    """
    target_date = pd.to_datetime(target_date).date()
    df['date'] = pd.to_datetime(df['timestamp']).dt.date

    training_dates = calendar_util.get_training_dates(
        target_date, lookback_days, lag=lag, frequency=model_train_freq
    )
    return df[df['date'].isin(training_dates)].copy()
```

```python
import pandas as pd
from joblib import Parallel, delayed
from tqdm import tqdm
import fwd_return


def _fwd_data_task(symbol, date, port_freq, fwd_return_interval, local_timezone):
    """
    Helper task to call fwd_return.fwd_data safely.
    """
    try:
        df = fwd_return.fwd_data(symbol, date, port_freq, fwd_return_interval, local_timezone)
        if df is not None and not df.empty:
            return df
    except Exception as e:
        print(f"[ERROR] fwd_data failed for {symbol} on {date}: {e}")
    return None


def create_multi_day_fwd_returns(
    symbols, 
    dates, 
    port_freq, 
    fwd_return_interval, 
    local_timezone="Asia/Kolkata", 
    nproc=4
):
    """
    Collect forward return datasets for multiple symbols and dates in parallel.
    Falls back to sequential processing if nproc == 1.
    """
    # Ensure symbols is a list
    if isinstance(symbols, str):
        symbols = [symbols]

    # Normalize dates to list of datetime.date
    if hasattr(dates, "__iter__") and not isinstance(dates, (str, bytes)):
        dates = [pd.to_datetime(d).date() for d in dates]
    else:
        dates = [pd.to_datetime(dates).date()]

    # Create task list
    tasks = [(symbol, date) for symbol in symbols for date in dates]
    print(f"[INFO] Loaded {len(tasks)} tasks for forward return computation.")

    results = []

    if nproc == 1:
        # Sequential execution with tqdm
        for symbol, date in tqdm(tasks, desc="Fetching fwd_return data (sequential)"):
            df = _fwd_data_task(symbol, date, port_freq, fwd_return_interval, local_timezone)
            if df is not None and not df.empty:
                results.append(df)
    else:
        # Parallel execution with tqdm
        results = Parallel(n_jobs=nproc, backend="loky")(
            delayed(_fwd_data_task)(symbol, date, port_freq, fwd_return_interval, local_timezone)
            for symbol, date in tqdm(tasks, desc="Fetching fwd_return data (parallel)")
        )
        # Filter valid DataFrames
        results = [df for df in results if df is not None and not df.empty]

    if not results:
        print("[WARN] No forward return data collected.")
        return pd.DataFrame()

    combined_df = pd.concat(results, ignore_index=True)
    print(f"[INFO] Combined fwd_return dataset contains {len(combined_df)} rows.")
    return combined_df


def merge_with_fwd_returns(
    final_prediction_df, 
    symbols, 
    dates, 
    port_freq, 
    fwd_return_interval, 
    local_timezone="Asia/Kolkata", 
    nproc=4
):
    """
    Wrapper: fetch forward returns and merge into prediction dataframe.
    """
    fwd_ret_df = create_multi_day_fwd_returns(
        symbols, dates, port_freq, fwd_return_interval, local_timezone, nproc=nproc
    )
    if fwd_ret_df.empty:
        print("[WARN] No fwd_return data to merge.")
        return final_prediction_df

    merged = final_prediction_df.merge(fwd_ret_df, on=["timestamp", "symbol"])
    print(f"[INFO] Final merged dataframe contains {len(merged)} rows.")
    return merged
```

# Feature Selection


```python
import numpy as np
import pandas as pd
from math import sqrt
from sklearn.linear_model import LinearRegression
from scipy.stats import norm
from joblib import Parallel, delayed

# ---------------------------
# Helpers
# ---------------------------

def _ensure_sorted(df):
    if not df.index.is_monotonic_increasing:
        df = df.sort_index()
    return df

def _split_into_parts(n, n_parts, min_points=100):
    """
    Return list of (start, end) index slices for n sequential points,
    using as many parts (<= n_parts) as possible such that each part has >= min_points.
    """
    n_parts = min(n_parts, max(1, n // min_points))  # upper bound by available data
    if n_parts <= 1:
        return [(0, n)]
    # Equal-ish chunks
    base = n // n_parts
    rem  = n % n_parts
    sizes = [base + (1 if i < rem else 0) for i in range(n_parts)]
    # If any chunk violates min_points, reduce n_parts and retry
    if min(sizes) < min_points:
        return _split_into_parts(n, n_parts - 1, min_points=min_points)
    bounds = []
    s = 0
    for sz in sizes:
        e = s + sz
        bounds.append((s, e))
        s = e
    return bounds

def _signals_from_predictions(pred):
    # +1 if > 0 else -1 (zero treated as -1 per spec)
    return np.where(pred > 0, 1, -1)

# ---------------------------
# Metrics
# ---------------------------

def sharpe_ratio(pnl):
    pnl = pd.Series(pnl).dropna()
    if len(pnl) < 2 or pnl.std(ddof=1) == 0:
        return np.nan
    return pnl.mean() / pnl.std(ddof=1) * sqrt(len(pnl))

def t_stat(pnl):
    pnl = pd.Series(pnl).dropna()
    n = len(pnl)
    if n < 2 or pnl.std(ddof=1) == 0:
        return np.nan
    return pnl.mean() / (pnl.std(ddof=1) / sqrt(n))

def probabilistic_sharpe_ratio(pnl, sr_benchmark=0.0):
    """
    Probabilistic Sharpe Ratio (PSR) from Bailey et al.
    PSR = Φ( ((SR_hat - SR*) * sqrt(n-1)) / sqrt(1 - γ3*SR_hat + ((γ4 - 1)/4)*SR_hat**2) )
    """
    r = pd.Series(pnl).dropna()
    n = len(r)
    if n < 2 or r.std(ddof=1) == 0:
        return np.nan
    # sample moments
    mu = r.mean()
    sig = r.std(ddof=1)
    sr_hat = mu / sig * sqrt(n)  # Sharpe defined like above (sqrt(n)*mean/std)
    # standardized central moments
    # unbiased sample skew/kurt (Fisher) for robustness
    g3 = r.skew()
    g4 = r.kurtosis(fisher=False)  # Pearson's kurtosis (3 = normal)
    denom = np.sqrt(max(1e-12, 1 - g3 * sr_hat + ((g4 - 1.0) / 4.0) * (sr_hat ** 2)))
    z = ((sr_hat - sr_benchmark) * np.sqrt(n - 1)) / denom
    return float(norm.cdf(z))

# ---------------------------
# Core: walk-forward LR -> signals -> PnL
# ---------------------------

def walkforward_lr_pnl(train_df, label_df, feature_cols, target_col, n_parts=5, min_points=100):
    """
    For X (train_df[feature_cols]) and y (label_df[target_col], aligned by index),
    split chronologically into parts, train LR on part i, predict on part i+1,
    convert preds to +/-1 signals, compute pnl = signal * y on the predicted slice.
    Returns: pnl Series aligned to the prediction slices (concatenated).
    """
    X = _ensure_sorted(train_df[feature_cols].copy())
    y = _ensure_sorted(label_df[[target_col]].copy())[target_col]

    # Align indices
    X, y = X.align(y, join='inner', axis=0)
    n = len(X)
    if n < max(2 * min_points, 200):
        # Need at least one train+one test part of 100 each -> suggest 200 minimum
        return pd.Series(dtype=float)

    parts = _split_into_parts(n, n_parts, min_points=min_points)
    if len(parts) < 2:
        return pd.Series(dtype=float)

    pnl_list = []
    for i in range(len(parts) - 1):
        tr_s, tr_e = parts[i]
        te_s, te_e = parts[i + 1]
        X_tr, y_tr = X.iloc[tr_s:tr_e], y.iloc[tr_s:tr_e]
        X_te, y_te = X.iloc[te_s:te_e], y.iloc[te_s:te_e]
        # Guard again
        if len(X_tr) < min_points or len(X_te) < min_points:
            continue

        model = LinearRegression()
        model.fit(X_tr.values, y_tr.values)
        preds = model.predict(X_te.values)
        sig = _signals_from_predictions(preds)
        pnl = sig * y_te.values
        pnl_series = pd.Series(pnl, index=y_te.index)
        pnl_list.append(pnl_series)

    if not pnl_list:
        return pd.Series(dtype=float)
    return pd.concat(pnl_list).sort_index()

# ---------------------------
# Feature Selection Orchestrator (Parallel)
# ---------------------------

def evaluate_feature_sets(train_df, label_df, target_col,
                          feature_sets=None, n_parts=5, min_points=100,
                          sr_benchmark=0.0, n_jobs=-1, backend="loky"):
    """
    Evaluate one or many feature sets in parallel.
    By default, evaluates each individual column in train_df as a separate set.

    Returns a DataFrame:
        columns = ['features', 'sharpe', 't_stat', 'psr', 'n_obs']
        plus a dict of pnl series per feature set.
    """
    train_df = _ensure_sorted(train_df)
    label_df = _ensure_sorted(label_df)

    if feature_sets is None:
        feature_sets = [[c] for c in train_df.columns.tolist()]

    def _eval(fs):
        pnl = walkforward_lr_pnl(train_df, label_df, fs, target_col,
                                 n_parts=n_parts, min_points=min_points)
        if pnl.empty:
            return {
                'features': tuple(fs),
                'sharpe': np.nan,
                't_stat': np.nan,
                'psr': np.nan,
                'n_obs': 0,
                'pnl': pnl
            }
        sr = sharpe_ratio(pnl)
        ts = t_stat(pnl)
        psr = probabilistic_sharpe_ratio(pnl, sr_benchmark=sr_benchmark)
        return {
            'features': tuple(fs),
            'sharpe': sr,
            't_stat': ts,
            'psr': psr,
            'n_obs': int(pnl.shape[0]),
            'pnl': pnl
        }

    results = Parallel(n_jobs=n_jobs, backend=backend)(
        delayed(_eval)(fs) for fs in feature_sets
    )

    # Tabular summary + pnl dict
    summary_rows = []
    pnl_dict = {}
    for r in results:
        summary_rows.append({
            'features': r['features'],
            'sharpe': r['sharpe'],
            't_stat': r['t_stat'],
            'psr': r['psr'],
            'n_obs': r['n_obs']
        })
        pnl_dict[r['features']] = r['pnl']

    summary_df = pd.DataFrame(summary_rows).sort_values(
        ['psr', 'sharpe', 'n_obs'], ascending=[False, False, False]
    ).reset_index(drop=True)

    return summary_df, pnl_dict

# ---------------------------
# Convenience API
# ---------------------------

class FeatureSelector:
    """
    High-level wrapper you can plug into your pipeline.
    """
    def __init__(self, n_parts=5, min_points=100, sr_benchmark=0.0, n_jobs=-1, backend="loky"):
        self.n_parts = n_parts
        self.min_points = min_points
        self.sr_benchmark = sr_benchmark
        self.n_jobs = n_jobs
        self.backend = backend

    def run(self, train_df, label_df, target_col, feature_sets=None):
        return evaluate_feature_sets(
            train_df=train_df,
            label_df=label_df,
            target_col=target_col,
            feature_sets=feature_sets,
            n_parts=self.n_parts,
            min_points=self.min_points,
            sr_benchmark=self.sr_benchmark,
            n_jobs=self.n_jobs,
            backend=self.backend
        )
```

```python
import numpy as np
import pandas as pd
from sklearn.base import BaseEstimator, TransformerMixin, RegressorMixin, clone
from sklearn.linear_model import Ridge
from sklearn.ensemble import BaggingRegressor
from lightgbm import LGBMRegressor
from sklearn.pipeline import Pipeline


# === Custom Transformer for Feature + Label Clipping ===
class OutlierClipper(BaseEstimator, TransformerMixin):
    def __init__(self, feature_cols, label_col=None, method="iqr", factor=1.5):
        self.feature_cols = feature_cols
        self.label_col = label_col
        self.method = method
        self.factor = factor
        self.boundaries_ = {}

    def _compute_bounds(self, values):
        if self.method == "iqr":
            q1, q3 = np.percentile(values, [25, 75])
            iqr = q3 - q1
            return q1 - self.factor * iqr, q3 + self.factor * iqr
        else:  # zscore
            mean, std = values.mean(), values.std()
            return mean - self.factor * std, mean + self.factor * std

    def fit(self, X, y=None):
        X = pd.DataFrame(X, columns=self.feature_cols)

        # Feature bounds
        for col in self.feature_cols:
            values = X[col].dropna()
            self.boundaries_[col] = self._compute_bounds(values)

        # Label bounds
        if y is not None and self.label_col is not None:
            values = pd.Series(y, name=self.label_col).dropna()
            self.boundaries_[self.label_col] = self._compute_bounds(values)

        return self

    def transform(self, X, y=None):
        X = pd.DataFrame(X, columns=self.feature_cols).copy()

        # Clip features
        for col in self.feature_cols:
            lower, upper = self.boundaries_[col]
            X[col] = X[col].clip(lower, upper)

        if y is not None and self.label_col is not None:
            lower, upper = self.boundaries_[self.label_col]
            y = pd.Series(y).clip(lower, upper).values
            return X.values, y

        return X.values


# === Custom Ensemble Wrapper ===
class EnsembleAvg(BaseEstimator, RegressorMixin):
    def __init__(self, models):
        self.models = models

    def fit(self, X, y):
        self.fitted_models_ = []
        for name, model in self.models:
            mdl = clone(model)
            mdl.fit(X, y)
            self.fitted_models_.append((name, mdl))
        return self

    def predict(self, X):
        preds = np.column_stack([mdl.predict(X) for _, mdl in self.fitted_models_])
        return preds.mean(axis=1)


# === Full Pipeline with Auto Label Handling ===
class ClippedPipeline(Pipeline):
    def fit(self, X, y=None, **fit_params):
        # First fit the clipper on both X and y
        clipper = self.named_steps["clipper"]
        clipper.fit(X, y)

        # Transform X and y
        X_clipped, y_clipped = clipper.transform(X, y)

        # Fit the rest of the pipeline on clipped data
        self.named_steps["ensemble"].fit(X_clipped, y_clipped)
        return self

    def predict(self, X):
        clipper = self.named_steps["clipper"]

        # Clip features
        X_clipped = clipper.transform(X)

        # Get raw predictions
        preds = self.named_steps["ensemble"].predict(X_clipped)

        # Clip predictions to training label bounds
        if clipper.label_col in clipper.boundaries_:
            lower, upper = clipper.boundaries_[clipper.label_col]
            preds = np.clip(preds, lower, upper)

        return preds


# === Model Builder ===
def ml_model(n_data_len, feature_cols, label_col):
    ridge_base = Ridge()
    ridge_bagging = BaggingRegressor(
        estimator=ridge_base,
        n_estimators=100,
        max_samples=0.7,
        max_features=0.5,
        bootstrap=True,
        n_jobs=-1,
        random_state=42
    )

    lgbm = LGBMRegressor(
        n_estimators=128, learning_rate=0.01, max_depth=-1, random_state=42,
        min_child_samples=int(0.1*n_data_len), min_child_weight=0.01, num_leaves=8, n_jobs=8
    )

    clipper = OutlierClipper(feature_cols, label_col, method="iqr", factor=1.5)

    pipeline = ClippedPipeline([
        ("clipper", clipper),
        ("ensemble", EnsembleAvg([
            ("Ridge_Bagging", ridge_bagging),
            ("LGBM", lgbm),
        ]))
    ])

    return pipeline
```

```python
import numpy as np
import pandas as pd
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error


forward_predictions = {}

CLASS_THRESH = 0  # if you later extend for classifiers

for train_date in model_train_dates:
    print(f"\nTraining models for {train_date}")

    # Build training dataset
    train_data = get_train_data(dataset_df, train_date, train_lookback_days, lag=1)
    train_data = train_data.replace([np.inf, -np.inf], np.nan).fillna(0)

    if train_data.empty:
        print("No training data found. Skipping...")
        continue

    X_train = train_data[feature_cols].values
    y_train = train_data[label_col].values

    if len(y_train) == 0:
        print("No training labels found. Skipping...")
        continue

    model = ml_model(len(train_data), feature_cols, label_col)
    # --- Fit pipeline (includes outlier clipping) ---
    model.fit(X_train, y_train)

    # In-sample prediction
    y_pred_train = model.predict(X_train)

    r2 = r2_score(y_train, y_pred_train)
    mse = mean_squared_error(y_train, y_pred_train)
    mae = mean_absolute_error(y_train, y_pred_train)

    print(f"In-sample R² Score: {r2:.10f}")
    print(f"In-sample MSE     : {mse:.10f}")
    print(f"In-sample MAE     : {mae:.10f}")

    # ----------------------
    # OOS / Forward Prediction
    # ----------------------
    pred_dates = model_prediction_dates[train_date]
    if pred_dates.empty:
        continue

    ASIA_TZ = 'Asia/Kolkata'
    pred_dates = pred_dates.tz_localize(ASIA_TZ)

    mask = (
        (dataset_df['timestamp'] >= pred_dates.min()) &
        (dataset_df['timestamp'] <= pred_dates.max())
    )
    pred_df = dataset_df[mask].copy()
    pred_df = pred_df.replace([np.inf, -np.inf], np.nan).fillna(0)

    if pred_df.empty:
        print("No data available for forward prediction window.")
        continue

    X_pred = pred_df[feature_cols].values
    y_true = pred_df[label_col] if label_col in pred_df else None

    # Predict using pipeline (clipping applied automatically with training bounds)
    y_pred_oos = model.predict(X_pred)
    pred_df["Prediction"] = y_pred_oos
    forward_predictions[train_date] = pred_df

    # Evaluate
    if y_true is not None and not y_true.isnull().all():
        r2_oos = r2_score(y_true, y_pred_oos)
        mse_oos = mean_squared_error(y_true, y_pred_oos)
        mae_oos = mean_absolute_error(y_true, y_pred_oos)
        print(f"OOS R² Score: {r2_oos:.10f}")
        print(f"OOS MSE     : {mse_oos:.10f}")
        print(f"OOS MAE     : {mae_oos:.10f}")
    else:
        print("Ground truth not available for OOS window.")
```
